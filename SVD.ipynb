```python
import numpy as np
```

```python
Nt = 10
h = np.random.randn(1, Nt) + 1j* np.random.randn(1, Nt)
```

```python
[u, sigma, vh] = np.linalg.svd(h, full_matrices=False)
```

```python
print(u, sigma, vh)
```

```python
print(h)
```

```python
print( u @ sigma @ vh)
```

```python
print( np.conjugate(np.transpose(h/np.linalg.norm(h))))
```

```python
print(np.conjugate(np.transpose(vh)))
```
